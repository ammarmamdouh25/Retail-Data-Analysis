## Quantity Validation

We check for negative and zero quantities because they may represent returns, cancellations, or invalid transactions.


```python
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Zero Quantity:", (df["Quantity"] == 0).sum())
```

```python
df["InvoiceDate"].dtype
```

```python
df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

df["InvoiceDate"].dtype
```

# 4. Data Cleaning

Based on the data quality assessment, the identified issues are handled carefully without removing valid business information unnecessarily.

```python
df = df.drop_duplicates()

print("Rows after removing duplicates:", df.shape[0])
```

```python
sales_df = df.copy()
```

```python
sales_df = sales_df[
    (sales_df["Quantity"] > 0) &
    (sales_df["UnitPrice"] > 0)
].copy()
```

```python
print("Rows after cleaning:", sales_df.shape[0])
```

# 5. Feature Engineering

New features are created from the existing data to support sales analysis and provide more useful business information.

```python
sales_df["Revenue"] = (
    sales_df["Quantity"] *
    sales_df["UnitPrice"]
)
```

```python
sales_df["Revenue"].sum()
```

```python
sales_df["InvoiceMonth"] = (
    sales_df["InvoiceDate"]
    .dt.to_period("M")
    .astype(str)
)
```

```python
sales_df["OrderType"] = np.where(
    sales_df["InvoiceNo"].astype(str).str.startswith("C"),
    "Cancellation",
    "Sale"
)
```

```python
sales_df.head()
```

# 7. Data Quality Summary

The dataset was examined for missing values, duplicate records, invalid quantities, invalid prices, and incorrect data types.

Duplicate records were removed to prevent double counting.

Negative quantities and non-positive prices were excluded from the sales analysis because they do not represent normal sales transactions.

Missing CustomerID values were retained because the transactions still contain useful sales information. However, these records should be excluded from customer-level analysis when customer identification is required.

InvoiceDate was converted to datetime format to support time-based analysis.

Two useful features were created: Revenue and InvoiceMonth. Revenue allows direct analysis of sales value, while InvoiceMonth supports monthly sales trend analysis.

After cleaning and validation, the sales dataset is suitable for further sales analysis.
